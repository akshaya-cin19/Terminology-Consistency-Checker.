```python

# Terminology Consistency Checker

import re
from collections import defaultdict

# Terms that may refer to the same concept
term_groups = {
    "Artificial Intelligence": [
        "AI", "Artificial Intelligence"
    ],
    "Machine Learning": [
        "ML", "Machine Learning"
    ],
    "User Interface": [
        "UI", "User Interface"
    ],
    "Database": [
        "DB", "Database"
    ],
    "Application Programming Interface": [
        "API", "Application Programming Interface"
    ]
}

def check_consistency(documents):
    usage = defaultdict(set)

    for doc_name, text in documents.items():
        for standard_term, variants in term_groups.items():
            for variant in variants:
                pattern = r"\b" + re.escape(variant) + r"\b"

                if re.search(pattern, text, re.IGNORECASE):
                    usage[standard_term].add(variant)

    print("=" * 50)
    print("TERMINOLOGY CONSISTENCY CHECKER")
    print("=" * 50)

    inconsistent_count = 0

    for standard_term, variants in usage.items():
        print(f"\nConcept: {standard_term}")
        print("Forms found:", ", ".join(sorted(variants)))

        if len(variants) > 1:
            inconsistent_count += 1
            print("Status: Multiple forms used")
            print("Suggestion: Use one preferred term consistently.")
        else:
            print("Status: Consistent in detected forms")

    print("\nConcepts with multiple forms:", inconsistent_count)
    print("=" * 50)


# Sample documents
documents = {
    "Report 1": (
        "AI supports the ML model. "
        "The User Interface displays the results."
    ),
    "Report 2": (
        "Artificial Intelligence uses Machine Learning. "
        "The UI displays data from the database."
    )
}

check_consistency(documents)
```
